```python
# == vs is with List
x=[10,20] #Python creates a list object in memory and x refers to that object.
y=[10,20] #Python creates another seprate list object. Although the values are the same, the two lists are different object.
          #Becouse list belongs to Mutable Type.
print(x==y) # Output: True  why? The == opeator compares the values/ content of two objects.
print(x is y) # Output: False Why? The is operater checks objects identity.
x.append(33) #The list referred to by x is modified. Only the x list changes because x and y refer to different list objects.
print(x==y) # Output: False because: x - [10,20,33] y - [10,20] Their contents are now different.
print(id(x))
print(id(y))
```

```python
#Student Performance Using Nested If
students={
    "Rahul":75,
    "Amit":32,
    "Priya":88,
    "Neha":39
}
for name, marks in students.items():
    if(marks>40):
        if(marks>=80):
            print(name,marks,"Excllent")
        elif(marks>=60):
            print(name,marks,"Good")
        elif(marks>=40):
            print(name,marks,"Average")
    else:
        print(name,marks,"Fail")
```

```python
#Employee Salary and Department
employees={
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"},
}
for name, data in employees.items():
    salary=data["salary"]
    department=data["department"]
    if(salary>=50000):
        if(department=="IT"):
            print(name, "Senior IT Employee")
        elif(department=="HR"):
            print(name, "Senior HR Employee")
        else:
            print(name, "Senior Employee")
    else:
        print(name, "Junior Employee")
```

```python
#Check Even/Odd Values
data={
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}
for lett, value in data.items():
    if value>20 and value%2==0:
        print(value, "Large Even")
    elif value<=20 and value%2==0:
        print(value, "Even")
    else:
        print(value, "Odd")
```

```python
#Shopping Cart Classification
products={
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200
}
for item, price in products.items():
    if price>=1000:
        if price>=5000:
            print(item, price, "Premium Product")
        else:
            print(item, price, "Expensive Product")
    else:
        print(item, price, "Affordable Product")
```

```python
#Student Eligibility
students={
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}
for name, data in students.items():
    att=data["attendance"]
    marks=data["marks"]
    if att>=75:
        if marks>=40:
            print(name,"Eligible")
        else:
            print(name,"Failed in Exam")
    else:
        print(name,"Not Eligible")
```

```python
#Bank Account Classification
accounts={
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}
for name, bal in accounts.items():
    if bal>0:
        if bal>=10000:
            print(name, "Premium Account")
        else:
            print(name, "Regural Account")
    elif bal==0:
        print(name, "No Balance")
    else:
        print(name, "Overdraft")
```

```python
employees={
    "Rahul": {"salary": 60000, "experience": 6},
    "Aman": {"salary": 45000, "experience": 4},
    "Priya": {"salary": 55000, "experience": 3},
    "Neha": {"salary": 40000, "experience": 2}
}
for name, data in employees.items():
    salary=data["salary"]
    experience=data["experience"]
    if salary>=50000:
        if experience>=5:
            print(name,"Senior")
        else:
            print(name,"Mid Level")
    else:
        if experience>=3:
            print(name,"Experienced Junior")
        else:
            print(name,"Junior/Fresher")
```

```python
n=5
for i in range(1,2*n):
    if i<=n: #i=1 n=5 i<=n, 1<=5
        star=1
    else:
        star= 2*n-i #
    space=n-star

    for j in range(0, space):
        print("  ", end=" ")

    for k in range(0,2*star-1):
        print("* ", end=" ")

    print()
```
